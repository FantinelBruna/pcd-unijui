```python
def calcular_maior_nota(notas):
    return max(notas)

def calcular_menor_nota(notas):
    return min(notas)

def calcular_media(notas):
    return sum(notas) / len(notas)

def contar_aprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota >= 6.0:
            quantidade += 1

    return quantidade

def contar_reprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota < 6.0:
            quantidade += 1

    return quantidade


quantidade_estudantes = int(input("Digite a quantidade de estudantes: "))

notas = []
for i in range(quantidade_estudantes):
    nota = float(input(f"Digite a nota do estudante {i + 1}: "))
    notas.append(nota)

maior_nota = calcular_maior_nota(notas)
menor_nota = calcular_menor_nota(notas)
media = calcular_media(notas)
aprovados = contar_aprovados(notas)
reprovados = contar_reprovados(notas)

print(f"Maior nota: {maior_nota:.2f}")
print(f"Menor nota: {menor_nota:.2f}")
print(f"Média da turma: {media:.2f}")
print(f"Estudantes aprovados: {aprovados}")
print(f"Estudantes reprovados: {reprovados}")
```

```python
def consultar_produto(produtos, codigo):
    if codigo in produtos:
        produto = produtos[codigo]

        print("Produto Encontrado:")
        print(f"Código: {codigo}")
        print(f"Nome: {produto['nome']}")
        print(f"Preço: R$ {produto['preco']:.2f}")
        print(f"Quantidade em estoque: {produto['quantidade']}")
    else:
        print("\nProduto não encontrado.")


def calcular_valor_total_estoque(produtos):
    valor_total = 0

    for produto in produtos.values():
        valor_total += produto["preco"] * produto["quantidade"]

    return valor_total


def encontrar_produto_maior_valor(produtos):
    produto_maior_valor = None
    maior_preco = -1

    for codigo, produto in produtos.items():
        if produto["preco"] > maior_preco:
            maior_preco = produto["preco"]

            produto_maior_valor = {
                "codigo": codigo,
                "nome": produto["nome"],
                "preco": produto["preco"]
            }

    return produto_maior_valor


def apresentar_resultados(produtos):
    valor_total = calcular_valor_total_estoque(produtos)
    produto_maior_valor = encontrar_produto_maior_valor(produtos)

    print(f"Valor total do estoque: R$ {valor_total:.2f}")
    print(f"Código: {produto_maior_valor['codigo']}")
    print(f"Nome: {produto_maior_valor['nome']}")
    print(f"Preço: R$ {produto_maior_valor['preco']:.2f}")


produtos = {}

quantidade_produtos = int(input("Digite a quantidade de produtos: "))

for i in range(quantidade_produtos):
    print(f"\nProduto {i + 1}")

    codigo = input("Código: ")
    nome = input("Nome: ")
    preco = float(input("Preço: R$ "))
    quantidade = int(input("Quantidade em estoque: "))

    produtos[codigo] = {
        "nome": nome,
        "preco": preco,
        "quantidade": quantidade
    }


codigo_consulta = input("\nDigite o código do produto que deseja consultar: ")

consultar_produto(produtos, codigo_consulta)
apresentar_resultados(produtos)
```
