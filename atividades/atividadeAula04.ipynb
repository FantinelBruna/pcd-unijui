1. Escreva um programa que leia um número e verifique se é ou não um número primo.
Para fazer essa verificação, calcule o resto da divisão do número por 2 e depois por todos os números ímpares até o número lido. Se o resto de uma dessas divisões for igual a zero, o número não é primo.
Observe que 0 e 1 não são primos e que 2 é o único número primo que é par.

```python
numero = int(input("Digite um número: "))

if numero == 0 or numero == 1:
    print("O número não é primo.")
elif numero == 2:
    print("O número é primo.")
elif numero % 2 == 0:
    print("O número não é primo.")
else:
    primo = True

    for divisor in range(3, numero, 2):
        if numero % divisor == 0:
            primo = False
            break

    if primo:
        print("O número é primo.")
    else:
        print("O número não é primo.")
```

Desafio:
Altere o programa 02 para, ao invés de verificar se um número específico é primo, a entrada é quantidade de números primos iniciais.
Exemplo: se a entrada for 5, apresentar a lista com os 5 primeiro numeros primos

```python
quantidade = int(input("Digite a quantidade de números primos: "))

contador = 0
numero = 2

while contador < quantidade:

    primo = True

    if numero == 2:
        primo = True

    elif numero % 2 == 0:
        primo = False

    else:
        for divisor in range(3, numero, 2):
            if numero % divisor == 0:
                primo = False
                break

    if primo:
        print(numero)
        contador += 1

    numero += 1
```

2)  Escreva um programa que calcule a raiz quadrada de um número, utilizando o método de Newton para obter um resultado aproximado.
Sendo n o número a obter a raiz quadrada.


Calcule p usando a fórmula p = (b+ (n/b) ) / 2.
  
Agora, calcule o quadrado de p.
A cada passo, faça b = p e recalcule p usando a fórmula apresentada.
Pare quando a diferença absoluta entre n e o quadrado de p for menor que 0,0001.

Ao final use a função SQRT (n) e mostre os dois resultados.

```python
import math

n = float(input("Digite um número: "))
b = n
p = (b + (n / b)) / 2

while abs(n - p ** 2) >= 0.0001:
    b = p
    p = (b + (n / b)) / 2

print("Raiz pelo método de Newton:", p)
print("Raiz usando sqrt:", math.sqrt(n))
```
