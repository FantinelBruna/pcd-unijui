Atividade I

Escreva um programa que receba do usuário os valores de um conjunto de características de um personagem inventado por você.
Leia os valores e grave um arquivo para posterior manipulação.
Escolha o formato de gravação.

O arquivo não precisa ser alterado, apenas sobreescrito.
Crie uma função para ler e apresentar (print) os dados gravados.

```python
import json

nome = input("Digite o nome da sereia: ")
cauda = input("Digite a cor da cauda: ")
poder = input("Digite o poder da sereia: ")

sereia = {
    "nome": nome,
    "cauda": cauda,
    "poder": poder
}

with open("sereia.json", "w") as arquivo:
    json.dump(sereia, arquivo, indent=4, ensure_ascii=False)

with open("sereia.json", "r") as arquivo:
  sereia = json.load(arquivo)

print("Nome:", sereia["nome"])
print("Cor da cauda:", sereia["cauda"])
print("Poder:", sereia["poder"])
```
