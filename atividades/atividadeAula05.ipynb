Atividade I

Escreva um programa que receba do usuário os valores de um conjunto de características de um personagem inventado por você.
Leia os valores e grave um arquivo para posterior manipulação.
Escolha o formato de gravação.

O arquivo não precisa ser alterado, apenas sobreescrito.
Crie uma função para ler e apresentar (print) os dados gravados.

```python
import json

nome = input("Digite o nome da sereia: ")
cauda = input("Digite a cor da cauda: ")
poder = input("Digite o poder da sereia: ")

sereia = {
    "nome": nome,
    "cauda": cauda,
    "poder": poder
}

with open("sereia.json", "w") as arquivo:
    json.dump(sereia, arquivo, indent=4, ensure_ascii=False)

with open("sereia.json", "r") as arquivo:
  sereia = json.load(arquivo)

print("Nome:", sereia["nome"])
print("Cor da cauda:", sereia["cauda"])
print("Poder:", sereia["poder"])
```

Atividade II

Gere uma lista com 20 números aleatórios e, para cada valor, calcule o resultado da multiplicação por 2 e por 10. Salve os dados em TXT, JSON, e CSV contendo três campos: valor original, valor *2, valor *10.

```python
import random
import json
import csv

dados = []

for i in range(20):
    numero = random.randint(1, 100)

    registro = {
        "valor_original": numero,
        "valor_x2": numero * 2,
        "valor_x10": numero * 10
    }

    dados.append(registro)


# txt
with open("numeros.txt", "w") as arquivo:
    arquivo.write("Valor Original | Valor * 2 | Valor * 10\n")

    for item in dados:
        arquivo.write(
            f"{item['valor_original']} | "
            f"{item['valor_x2']} | "
            f"{item['valor_x10']}\n"
        )

print("\n\nconteudo txt:\n")
with open("numeros.txt", "r") as arquivo:
    conteudo = arquivo.read()
    print(conteudo)


# json
with open("numeros.json", "w") as arquivo:
    json.dump(dados, arquivo, indent=4)

print("\n\nconteudo json:\n")
with open("numeros.json", "r") as arquivo:
    conteudo_json = json.load(arquivo)

    print(json.dumps(conteudo_json, indent=4))

# csv
with open("numeros.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)

    escritor.writerow([
        "valor_original",
        "valor_x2",
        "valor_x10"
    ])

    for item in dados:
        escritor.writerow([
            item["valor_original"],
            item["valor_x2"],
            item["valor_x10"]
        ])

print("\n\nconteudo csv:\n")
with open("numeros.csv", "r", encoding="utf-8") as arquivo:
    leitor = csv.reader(arquivo)

    for linha in leitor:
        print(linha)
```
