### 1. Dados

```python
import numpy as np

dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda",
    "Terça",
    "Quarta",
    "Quinta",
    "Sexta",
    "Sábado",
    "Domingo"
])

indicadores = np.array([
    "Pedidos",
    "Faturamento",
    "Clientes",
    "Ticket Médio",
    "Reclamações"
])

print(dados)
```

### Parte 1. Conhecendo os dados

```python
print("Quantidade de linhas:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Quantidade total de valores:", dados.size)
print("Número de dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)
```

### Parte 2. Acessando os dados

```python
print("Dados de sexta-feira:")
print(dados[4])

print("\nFaturamento:")
print(dados[:, 1])

print("\nReclamações:")
print(dados[:, 4])

print("\nDados de quinta-feira até domingo:")
print(dados[3:])

print("\nPedidos, faturamento e clientes:")
print(dados[:, 0:3])
```

### Parte 3. Operações numéricas

```python
total_pedidos = np.sum(dados[:, 0])
total_faturamento = np.sum(dados[:, 1])
total_clientes = np.sum(dados[:, 2])

print("Total de pedidos:", int(total_pedidos))
print(f"Faturamento total: R$ {total_faturamento:.2f}")
print("Total de clientes:", int(total_clientes))

media_pedidos = np.mean(dados[:, 0])
media_faturamento = np.mean(dados[:, 1])
media_reclamacoes = np.mean(dados[:, 4])

print(f"\nMédia diária de pedidos: {media_pedidos:.2f}")
print(f"Média diária de faturamento: R$ {media_faturamento:.2f}")
print(f"Média diária de reclamações: {media_reclamacoes:.2f}")
```

### Parte 4. Análise por dia

```python
maior_pedidos = np.argmax(dados[:, 0])
menor_pedidos = np.argmin(dados[:, 0])

print("Dia com mais pedidos:", dias[maior_pedidos])
print("Quantidade:", int(dados[maior_pedidos, 0]))

print("\nDia com menos pedidos:", dias[menor_pedidos])
print("Quantidade:", int(dados[menor_pedidos, 0]))

maior_faturamento = np.argmax(dados[:, 1])
menor_faturamento = np.argmin(dados[:, 1])

print("\nDia com maior faturamento:", dias[maior_faturamento])
print(f"Valor: R$ {dados[maior_faturamento, 1]:.2f}")

print("\nDia com menor faturamento:", dias[menor_faturamento])
print(f"Valor: R$ {dados[menor_faturamento, 1]:.2f}")

amplitude = np.max(dados[:, 1]) - np.min(dados[:, 1])

print(f"\nAmplitude do faturamento: R$ {amplitude:.2f}")
```

### Parte 5. Máscaras e filtros

```python
print("Dias com faturamento superior a R$ 12.000:")
print(dias[dados[:, 1] > 12000])

print("\nDias com mais de 150 pedidos:")
print(dias[dados[:, 0] > 150])

print("\nDias com mais de 4 reclamações:")
print(dias[dados[:, 4] > 4])

print("\nFaturamento dos dias com mais de 150 pedidos:")
print(dados[dados[:, 0] > 150, 1])

abaixo_media = dados[:, 1] < np.mean(dados[:, 1])

print("\nQuantidade de dias abaixo da média:")
print(np.sum(abaixo_media))
```

### Parte 6. Classificação dos dias

```python
classificacao = np.where(
    dados[:, 1] >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

print("Classificação dos dias:")
print(np.column_stack((dias, classificacao)))

quantidade_alto = np.sum(classificacao == "Alto desempenho")

print("\nDias de alto desempenho:", quantidade_alto)
```

### Parte 7. Estatística

```python
faturamento = dados[:, 1]

media = np.mean(faturamento)
mediana = np.median(faturamento)
desvio_padrao = np.std(faturamento)
minimo = np.min(faturamento)
maximo = np.max(faturamento)

print(f"Média: R$ {media:.2f}")
print(f"Mediana: R$ {mediana:.2f}")
print(f"Desvio padrão: R$ {desvio_padrao:.2f}")
print(f"Mínimo: R$ {minimo:.2f}")
print(f"Máximo: R$ {maximo:.2f}")
```

### Parte 8. Trabalhando com axis

```python
medias = np.mean(dados, axis=0)

print("Média de cada indicador:")
print(np.round(medias, 2))

maiores = np.max(dados, axis=0)

print("\nMaior valor de cada indicador:")
print(maiores)

menores = np.min(dados, axis=0)

print("\nMenor valor de cada indicador:")
print(menores)

desvios = np.std(dados, axis=0)

variacao_relativa = (desvios / medias) * 100

print("\nVariação relativa de cada indicador (%):")
print(np.round(variacao_relativa, 2))

maior_variacao = np.argmax(variacao_relativa)

print("\nIndicador com maior variação relativa:")
print(indicadores[maior_variacao])
```

### Parte 9. Análise integrada

```python
melhor_dia = np.argmax(dados[:, 1])

print("Melhor dia da semana:", dias[melhor_dia])

dia_mais_pedidos = np.argmax(dados[:, 0])

print("\nTambém apresentou mais pedidos?")
print("Sim" if melhor_dia == dia_mais_pedidos else "Não")

dia_mais_reclamacoes = np.argmax(dados[:, 4])

print("\nTambém apresentou mais reclamações?")
print("Sim" if melhor_dia == dia_mais_reclamacoes else "Não")

filtro = (dados[:, 0] > 150) & (dados[:, 1] > 20000)

print("\nQuantidade de dias que atendem às duas condições:")
print(np.sum(filtro))

print("Dias:", dias[filtro])
```

### Parte 10. Desafio de análise

```python
total_pedidos = np.sum(dados[:, 0])
total_faturamento = np.sum(dados[:, 1])

media_pedidos = np.mean(dados[:, 0])
media_faturamento = np.mean(dados[:, 1])

melhor_dia = dias[np.argmax(dados[:, 1])]
pior_dia = dias[np.argmin(dados[:, 1])]

maior_reclamacoes = np.max(dados[:, 4])
dia_reclamacoes = dias[np.argmax(dados[:, 4])]

dias_alto_desempenho = dias[dados[:, 1] >= 20000]

print("=============================================")
print("RESUMO DA SEMANA")
print("=============================================")

print(f"Total de pedidos: {int(total_pedidos)}")
print(f"Faturamento total: R$ {total_faturamento:.2f}")
print(f"Média diária de pedidos: {media_pedidos:.2f}")
print(f"Média diária de faturamento: R$ {media_faturamento:.2f}")

print(f"Melhor dia: {melhor_dia}")
print(f"Pior dia: {pior_dia}")

print(f"Maior número de reclamações: {int(maior_reclamacoes)}")
print(f"Dia com mais reclamações: {dia_reclamacoes}")

print(f"Dias de alto desempenho: {len(dias_alto_desempenho)}")
print("Dias:", ", ".join(dias_alto_desempenho))

print("=============================================")
```

### Parte 11. Conclusão

A partir da análise dos dados, foi possível identificar que a empresa realizou 948 pedidos durante a semana, alcançando um faturamento total de R$ 124.000,00, com uma média diária de R$ 17.714,29.

Os melhores resultados foram registrados na quinta-feira, sexta-feira e sábado, dias classificados como de alto desempenho por apresentarem faturamento igual ou superior a R$ 20.000,00. A sexta-feira destacou-se como o melhor dia da semana, com 175 pedidos e um faturamento de R$ 24.500,00.

A quarta-feira merece atenção por apresentar a menor quantidade de pedidos e o menor faturamento, de R$ 11.700,00. Seria interessante investigar os fatores que contribuíram para esse resultado e avaliar possíveis estratégias para melhorar as vendas nesse dia.

Isso indica uma possível relação entre o aumento da demanda e a quantidade de reclamações, embora os dados de apenas uma semana não sejam suficientes para confirmar essa relação.
