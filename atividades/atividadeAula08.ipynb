Escrever um programa usando Selenium para Scrap da temperatura (mínima e máxima) do dia, de uma cidade qualquer (informada pelo usuário), de algum site de previsão do tempo.

```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
from urllib.parse import quote
import json
```

```python
cidade = input("Digite o nome da cidade: ")

navegador = gs.Chrome()

cidade_url = quote(cidade)
url = f"https://wttr.in/{cidade_url}?format=j1"

navegador.get(url)
sleep(3)

conteudo = navegador.find_element(By.TAG_NAME,"body").text

dados = json.loads(conteudo)
hoje = dados["weather"][0]

temperatura_maxima = hoje["maxtempC"]
temperatura_minima = hoje["mintempC"]

print("\n===== PREVISÃO DO TEMPO =====")
print("Cidade:", cidade)
print("Temperatura máxima:", temperatura_maxima, "°C")
print("Temperatura mínima:", temperatura_minima, "°C")

navegador.quit()
```
