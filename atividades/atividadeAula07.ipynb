Usandos os métodos do Pandas, localize a sua cidade de residência e:
- mostre todas as informações
- qual o mês de temperatura mínima
- qual o mês de maior déficit hídrico
- a média de temperatura pelas 4 estações do ano

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')

cidade = df[
    (df['nome_municipio'] == 'Horizontina') &
    (df['uf'] == 'RS')
]

print("===== DADOS DE HORIZONTINA =====")
display(cidade.T)


colunas_tmin = [
    'tmin_jan', 'tmin_fev', 'tmin_mar', 'tmin_abr',
    'tmin_mai', 'tmin_jun', 'tmin_jul', 'tmin_ago',
    'tmin_set', 'tmin_out', 'tmin_nov', 'tmin_dez'
]

mes_temp_minima = cidade[colunas_tmin].idxmin(axis=1).iloc[0]
valor_temp_minima = cidade[colunas_tmin].min(axis=1).iloc[0]

mes_temp_minima = mes_temp_minima.replace('tmin_', '')

print("\n===== TEMPERATURA MÍNIMA =====")
print("Mês:", mes_temp_minima.capitalize())
print(f"Temperatura mínima: {valor_temp_minima:.2f} °C")


colunas_def = [
    'def_jan', 'def_fev', 'def_mar', 'def_abr',
    'def_mai', 'def_jun', 'def_jul', 'def_ago',
    'def_set', 'def_out', 'def_nov', 'def_dez'
]

mes_maior_deficit = cidade[colunas_def].idxmax(axis=1).iloc[0]
valor_maior_deficit = cidade[colunas_def].max(axis=1).iloc[0]

mes_maior_deficit = mes_maior_deficit.replace('def_', '')

print("\n===== MAIOR DÉFICIT HÍDRICO =====")
print("Mês:", mes_maior_deficit.capitalize())
print(f"Déficit hídrico: {valor_maior_deficit:.2f} mm/mês")


verao = cidade[
    ['temp_media_dez', 'temp_media_jan', 'temp_media_fev']
].mean(axis=1).iloc[0]

outono = cidade[
    ['temp_media_mar', 'temp_media_abr', 'temp_media_mai']
].mean(axis=1).iloc[0]

inverno = cidade[
    ['temp_media_jun', 'temp_media_jul', 'temp_media_ago']
].mean(axis=1).iloc[0]

primavera = cidade[
    ['temp_media_set', 'temp_media_out', 'temp_media_nov']
].mean(axis=1).iloc[0]


print("\n===== MÉDIA POR ESTAÇÃO =====")
print(f"Verão: {verao:.2f} °C")
print(f"Outono: {outono:.2f} °C")
print(f"Inverno: {inverno:.2f} °C")
print(f"Primavera: {primavera:.2f} °C")
```
