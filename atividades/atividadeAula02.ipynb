1. Implemente um programa que converta o valor de uma velocidade média em km/h para m/s. Para isso, o usuário deve informar o valor da velocidade média. Sabe-se que o fator utilizado para essa conversão é 3,6.


```python
velocidadeKmH = float (input('Qual a velocidade média (em km/h):'))
velocidadeMs = velocidadeKmH / 3.6
print(f'A velocidade em m/s é {velocidadeMs:.2f}')
```

2. Construa um programa que tem como dados de entrada dois pontos quaisquer no plano cartesiano: P1 e P2. Considere que P1 é definido pelas coordenadas x1 e y1, enquanto P2 por x2 e y2 . O programa deve calcular e escrever a distância entre os pontos P1 e P2.

A fórmula que calcula a distância entre os dois pontos é dada por:
 dAB = raiz quadrada ( (xB – xA)² + (yB – yA)² ).

Para receber os dados de entrada, use as variáveis x1, y1, x2 e y2.
Dica: A função que calcula a raiz quadrada de um número real é a sqrt (square root).
Para usá-la, importe-a da biblioteca math da seguinte forma: from math import sqrt.

```python
from math import sqrt


x1 = int (input('Informe a cordenada X do ponto 1 assim:'))
y1 = int (input('Informe a cordenada y do ponto 1 assim:'))
x2 = int (input('Informe a cordenada X do ponto 2 assim:'))
y2 = int (input('Informe a cordenada y do ponto 2 assim:'))


parteX = (x2 - x1) ** 2
parteY = (y2 - y1) ** 2
soma = parteX + parteY
total = sqrt(soma)


print(f'A distância entre os pontos P1[{x1},{y1}] e P2 [{x2},{y2}] é {total:.2f}')
```

3. Uma imobiliária paga aos seus corretores um salário base de R$ 2.500,00. Além disso, uma comissão de R$ 500,00 por cada imóvel vendido e 5% do valor de cada venda. Construa um programa que solicite o nome do corretor, a quantidade de imóveis vendidos e o valor total de suas vendas. Ao fim, o programa deve calcular e escrever o salário final do corretor de imóveis.

```python
nome = str (input('Informe o nome do vendedor:'))
qtdeVendas = int (input('Informe a quantidade de imóveis vendidos:'))
vlrTotal = float (input('Inform o valor total das vendas:'))


salarioBase = 2500
comissaoQtde = 500 * qtdeVendas
comissaoValor = vlrTotal * 0.05


salario = salarioBase + comissaoQtde + comissaoValor


print(f'O salário do corretor {nome} é de R$ {salario:.2f}')
```
