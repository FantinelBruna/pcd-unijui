```python
import pandas as pd

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv', sep=';')
print(f'\n {df}')
```

```python
def classificar_vinho(nota):
    if nota <= 4:
        return 'Baixa'
    elif nota <= 6:
        return 'Média'
    else:
        return 'Alta'

df['Class'] = df['quality'].apply(classificar_vinho)
print(f'\n {df}')
```

```python
x = df[df.columns.difference(['quality', 'Class'])].values
print(f'x: {x}')

y = df['Class'].values
print(f'y: {y}')
```

```python
from sklearn.neighbors import KNeighborsClassifier

vinho_classificador = KNeighborsClassifier(n_neighbors = 3)
vinho_classificador.fit(x, y)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(vinho_classificador, x, y, scoring='accuracy', cv=5)
print(f'Acurácia: {scores.mean()}')
```

```python
for i in range(1, 11):
    vinho_classificador = KNeighborsClassifier(n_neighbors=i)
    vinho_classificador.fit(x, y)

    scores = cross_val_score(vinho_classificador, x, y, scoring='accuracy', cv=5)

    print(f'Vizinhos: {i} | Acurácia: {scores.mean()}')
```

```python
vinho_classificador = KNeighborsClassifier(n_neighbors=7)
vinho_classificador.fit(x, y)
```

```python
teste = [x[0]]

print(f'Classificação prevista: {vinho_classificador.predict(teste)}')
print(f'Classificação real: {y[0]}')
```
