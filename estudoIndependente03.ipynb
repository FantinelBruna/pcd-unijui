# Web Scraping de Livros com Python

**Bruna Fantinel**

**Objetivo:** Desenvolver uma aplicação em Python para coletar automaticamente informações de livros disponíveis em um site da Web.

**Fonte dos dados:** https://books.toscrape.com/

**Descrição:** O projeto utiliza técnicas de Web Scraping para acessar páginas de um catálogo de livros e extrair informações como título, preço, avaliação, disponibilidade e link. Os dados são organizados com a biblioteca Pandas e armazenados em um arquivo CSV para utilização posterior.

**Bibliotecas utilizadas:** Requests, BeautifulSoup e Pandas.

1. Importação das bibliotecas

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
```

2. Acessando o site

```python
url = "https://books.toscrape.com/"

resposta = requests.get(url, timeout=15)
resposta.raise_for_status()

print("Status da requisição:", resposta.status_code)

soup = BeautifulSoup(resposta.text, "html.parser")

print("Título da página:", soup.title.text)
```

3. Coletando os dados dos livros

```python

livros = []

for pagina in range(1, 4):

    url = f"https://books.toscrape.com/catalogue/page-{pagina}.html"

    resposta = requests.get(url, timeout=15)
    resposta.raise_for_status()
    soup = BeautifulSoup(resposta.text, "html.parser")

    produtos = soup.find_all("article", class_="product_pod")

    for produto in produtos:

        titulo = produto.h3.a["title"]

        preco = produto.find("p", class_="price_color").text
        preco = float(preco.replace("Â£", ""))

        avaliacao = produto.find("p", class_="star-rating")["class"][1]

        disponibilidade = produto.find("p", class_="instock availability").text.strip()

        link = produto.h3.a["href"]
        link = "https://books.toscrape.com/catalogue/" + link

        livros.append({
            "Titulo": titulo,
            "Preco_GBP": preco,
            "Avaliacao": avaliacao,
            "Disponibilidade": disponibilidade,
            "Link": link
        })

    print(f"Página {pagina} coletada: {len(produtos)} livros")

    time.sleep(1)

print("Total de livros coletados:", len(livros))
```

4. Organizando os dados em uma tabela

```python
df = pd.DataFrame(livros)
df.head(10)
```

5. Salvando os dados em CSV

```python
df.to_csv("livros.csv", index=False, encoding="utf-8-sig")
print("Arquivo CSV gerado com sucesso!")
```

6. Conferindo os dados salvos

```python
dados = pd.read_csv("livros.csv")

print("Quantidade de livros:", len(dados))
print("Colunas disponíveis:", dados.columns.tolist())

dados.head()
```

7. Exploração dos dados

```python
print("Preço médio dos livros:", round(df["Preco_GBP"].mean(), 2))

print("Livro mais caro:")
print(df.loc[df["Preco_GBP"].idxmax(), ["Titulo", "Preco_GBP"]])

print("Quantidade por avaliação:")
print(df["Avaliacao"].value_counts())
```
