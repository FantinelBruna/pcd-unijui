Público: Sem prefixo especial. O atributo ou método pode ser acessado
de qualquer lugar, dentro ou fora da classe.

Protegido: Usa um sublinhado simples no início (_). É um aviso por
convenção para indicar que o membro deve ser usado apenas dentro da
classe ou de subclasses (herança).

Privado: Usa dois sublinhados no início (__). O Python altera
internamente o nome do atributo para _NomeDaClasse__atributo),
dificultando o acesso acidental fora da classe.

```python
# Definindo uma classe Pet e seus atributos
# Já estamos usando Abstração ao definir a classe
class Pet:
    # método construtor
    def __init__(self, nome: str, idade: int, peso: float, cor: str, tipo:str, __castrado=False):
        self.nome = nome
        self.idade = idade
        self.peso = peso
        self.cor = cor
        self.__tipo = tipo

        # a restrição de acesso aos atributos é definida ao inserir dois underscores ('__') ao atributo
        self.__castrado = __castrado

    def setTipo(self, tipo: str):
        self.__tipo = tipo

    def getTipo(self):
        return self.__tipo

    def setCor(self, cor: str):
        self.cor = cor

    def getCor(self):
        return self.cor

    def setCastracao(self):
        self.__anestesia()
        self.__castrado = True

    def getCastracao(self):
        return self.__castrado

    def apresentacao(self):
        print(f'Olá, meu nome é {self.nome}. Tenho {self.idade} anos, minha cor é {self.cor} e peso {self.peso} kg.')

    # a restrição de acesso aos métodos é definida ao inserir dois underscores ('__') antes do nome
    def __anestesia(self):
        dose = self.peso / 1000

    # força erro se não for implementado nas classes filhas
    def tipoComida(self):
        raise NotImplementedError()

    # para declarar um método sem conteúdo (vazio) temos que usar o comando 'pass' (já que o python não tem limitadores de blocos)
    def altura(self):
        pass

    #O __repr__ serve para exibir o objeto para o programador, usada pelo console do Python
    def __repr__(self):
        return (
            f"{self.__class__.__name__}"
        )

    #O __str__ serve para exibir o objeto para usuário final, usada pelo comando print e pela função str
    #def __str__(self):
    #    return (
    #        f"{self.__class__.__name__} - Olá, meu nome é {self.nome}. Tenho {self.idade} anos, minha cor é {self.cor} e peso {self.peso} kg."
    #    )
```

Inheritance - New Constructor

```python
# Definindo uma classe Pet e seus atributos
# Já estamos usando Abstração ao definir a classe

class Passaro(Pet):
    # método construtor
    def __init__(self, nome: str, idade: int, peso: float, cor: str, bico: str):
        super().__init__(nome, idade, peso, cor, 'Passaro')
        self.bico = bico

    def setBico(self, bico: str):
        self.bico = bico

    def getBico(self):
        return self.bico

    def tipoComida(self):
        return (
            "Alpiste"
        )
```

Using a Subclass

```python
cardeal = Passaro(nome='Pardal', idade=2, peso=0.7, cor='Amarela', bico='Fino')

print(cardeal.nome)
print(cardeal.bico)
print(cardeal.getCor())
print(cardeal.getTipo())
print(cardeal.tipoComida())
```
