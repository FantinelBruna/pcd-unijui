```python
class Calculadora:

    def somar(self, numero1, numero2):
        return numero1 + numero2


    def subtrair(self, numero1, numero2):
        return numero1 - numero2


    def multiplicar(self, numero1, numero2):
        return numero1 * numero2


    def dividir(self, numero1, numero2):
        if numero2 == 0:
            return "Erro: divisão por zero"

        return numero1 / numero2


    def raiz_quadrada(self, numero):
      if numero < 0:
          return "Erro: número negativo"

      if numero == 0:
          return 0

      b = numero
      p = (b + (numero / b)) / 2

      while abs(numero - p ** 2) >= 0.0001:
          b = p
          p = (b + (numero / b)) / 2

      return p

#------------------------------------------------------------------

calculadora = Calculadora()

opcao = ""

while opcao != "0":
    print("\n--- CALCULADORA ---")
    print("1 - Soma")
    print("2 - Subtração")
    print("3 - Multiplicação")
    print("4 - Divisão")
    print("5 - Raiz quadrada")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        numero1 = float(input("Digite o primeiro número: "))
        numero2 = float(input("Digite o segundo número: "))

        resultado = calculadora.somar(numero1, numero2)
        print("Resultado:", resultado)

    elif opcao == "2":
        numero1 = float(input("Digite o primeiro número: "))
        numero2 = float(input("Digite o segundo número: "))

        resultado = calculadora.subtrair(numero1, numero2)
        print("Resultado:", resultado)

    elif opcao == "3":
        numero1 = float(input("Digite o primeiro número: "))
        numero2 = float(input("Digite o segundo número: "))

        resultado = calculadora.multiplicar(numero1, numero2)
        print("Resultado:", resultado)

    elif opcao == "4":
        numero1 = float(input("Digite o primeiro número: "))
        numero2 = float(input("Digite o segundo número: "))

        resultado = calculadora.dividir(numero1, numero2)
        print("Resultado:", resultado)

    elif opcao == "5":
        numero = float(input("Digite um número: "))
        resultado = calculadora.raiz_quadrada(numero)
        print("Resultado:", resultado)

    elif opcao != "0":
        print("Opção inválida.")
```
