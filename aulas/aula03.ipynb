```python
# Etapa 1 — Entrada e cálculo, o programa deve:
# Criar três variáveis para armazenar as notas, inicializar com valores fixos.
# Calcular a média das três notas.
# Exibir as notas e a média.

a = 2
b = 4
c = 6

media = (a + b + c) / 3

print(f'A média das notas ({a}, {b}, {c}) é {media:.2f}')
```

```python
# Etapa 2 — Melhorando o programa
# Modificar o programa para:
# Receber o do aluno usando input().
# Receber as três notas usando input().
# Converter as notas para float.
# Calcular a média.
# Exibir um pequeno relatório.

nome = (input('Informe o nome do aluno:'))

a = float (input('Informe o primeira nota do aluno:'))
b = float (input('Informe o segunda nota do aluno:'))
c = float (input('Informe o terceira nota do aluno:'))

media = (a + b + c) / 3

print('=========')
print('Aluno: ' + nome)
print('')
print(f'Primeira nota: {a}')
print(f'Segunda nota: {b}')
print(f'Terceira nota: {b}')
print('')
print(f'Média das notas: {media}')
print('=========')
```
