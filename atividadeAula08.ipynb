```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
from selenium.webdriver.chrome.options import Options

options = Options()

options.add_experimental_option(
    "prefs",
    {
        "profile.default_content_setting_values.geolocation": 2
    }
)

navegador = gs.Chrome(options=options)
```

```python
url = "https://www.climatempo.com.br"

navegador.get(url)
sleep(2)

print("Página:", navegador.title)
```

```python
cidade = input("Digite a cidade: ")

campos = navegador.find_elements(
    By.XPATH,
    "//input[contains(@placeholder, 'cidade')]"
)

for campo in campos:
    if campo.is_displayed():
        campo.send_keys(cidade)
        break

sleep(2)
```

```python
# Localizando a cidade nos resultados
resultados = navegador.find_elements(
    By.XPATH,
    "//li[contains(@class, 'autocomplete-item')]//a"
)

for resultado in resultados:
    if resultado.is_displayed() and cidade.split(" - ")[0] in resultado.text:
        resultado.click()
        break

sleep(3)

# Conferindo a página
print("\nPágina após o clique:")
print(navegador.title)
print(navegador.current_url)

# Localizando a previsão de hoje
previsao = navegador.find_element(
    By.XPATH,
    "//*[contains(text(), 'temperaturas entre')]"
).text

print("\nPrevisão encontrada:")
print(previsao)

# Extraindo as temperaturas
temperaturas = re.findall(r"(\d+)°", previsao)

minima = temperaturas[0]
maxima = temperaturas[1]

# Exibindo o resultado
print("\nTEMPERATURAS DE HOJE:")
print("Cidade:", cidade)
print("Temperatura mínima:", minima + "°C")
print("Temperatura máxima:", maxima + "°C")

# Finalizando o navegador
navegador.quit()
```
