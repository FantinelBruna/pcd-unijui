Atividade I

Escreva um programa em Python que gere um número aleatório de 0 a 15, usando a biblioteca 'random' e o método 'randint()'. Na sequência, utilizando um loop faça com que o usuário tente acertar o número gerado.
A cada vez que o número é acertado, armazene num arquivo o número gerado e a quantidade de tentativas que o usuário realizou para acertar. Quando o usuário decidir parar de adivinhar, mostre todo o histórico que ele construiu durante a execução do programa.

```python
import random

historico = []

while True:
    numero_sorteado = random.randint(0, 15)
    tentativas = 0

    print("\nUm número entre 0 e 15 foi sorteado.")

    while True:
        palpite = int(input("Digite seu palpite: "))
        tentativas += 1

        if palpite == numero_sorteado:
            print(f"Você acertou em {tentativas} tentativa(s)!")
            break
        else:
            print("Você errou. Tente novamente.")

    historico.append({
        "numero": numero_sorteado,
        "tentativas": tentativas
    })

    with open("historico.txt", "a", encoding="utf-8") as arquivo:
        arquivo.write(
            f"Número sorteado: {numero_sorteado} | "
            f"Tentativas: {tentativas}\n"
        )

    continuar = input("\nDeseja jogar novamente? (s/n): ").lower()

    if continuar != "s":
        break


print("\n===== HISTÓRICO =====")

for item in historico:
    print(
        f"Número sorteado: {item['numero']} | "
        f"Tentativas: {item['tentativas']}"
    )
```

Atividade II

Crie um programa que permita adicionar itens para uma lista de compras.
O usuário vai adicionando a quantidade e o valor de cada item da lista. Ao final, deve mostrar o valor total das compras e salvar a lista num arquivo .CSV.

```python
import csv

lista_compras = []

while True:
    print("\n===== LISTA DE COMPRAS =====")

    produto = input("Digite o nome do produto: ")
    quantidade = int(input("Digite a quantidade: "))
    valor = float(input("Digite o valor unitário: R$ "))

    total_item = quantidade * valor

    lista_compras.append({
        "produto": produto,
        "quantidade": quantidade,
        "valor_unitario": valor,
        "total": total_item
    })

    continuar = input("\nDeseja adicionar outro item? (s/n): ").lower()

    if continuar != "s":
        break


valor_total = 0
for item in lista_compras:
    valor_total += item["total"]


print("\n===== COMPRA =====")

for item in lista_compras:
    print(
        f"{item['produto']} | "
        f"Quantidade: {item['quantidade']} | "
        f"Valor unitário: R$ {item['valor_unitario']:.2f} | "
        f"Total: R$ {item['total']:.2f}"
    )

print(f"\nValor total da compra: R$ {valor_total:.2f}")


with open("lista_compras.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)

    escritor.writerow([
        "Produto",
        "Quantidade",
        "Valor Unitário",
        "Total"
    ])

    for item in lista_compras:
        escritor.writerow([
            item["produto"],
            item["quantidade"],
            item["valor_unitario"],
            item["total"]
        ])

    escritor.writerow([
        "",
        "",
        "Valor Total",
        valor_total
    ])


print("\nLista salva no arquivo lista_compras.csv")
```
